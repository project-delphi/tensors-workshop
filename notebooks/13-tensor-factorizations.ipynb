## What this deep dive covers / Qué cubre este estudio a fondo

Four tensor models, fitted to the same real order-3 tensor:

| Model | Shape of the answer |
|---|---|
| **CP** | A sum of `R` rank-one tensors. Unique under mild conditions. |
| **Tucker** | A small core, plus one factor matrix per mode. |
| **Tensor train** | A chain of order-3 carriages, linked by bond dimensions. |
| **Block term** | A sum of small Tucker blocks — CP and Tucker as two extremes of one family. |

> 🇪🇸 Cuatro modelos tensoriales sobre el mismo tensor real de orden 3: **CP**, una suma de `R` tensores de rango uno, única bajo condiciones suaves; **Tucker**, un núcleo pequeño más una matriz de factores por modo; **tensor train**, una cadena de vagones de orden 3 unidos por dimensiones de enlace; y **términos por bloques**, una suma de bloques Tucker pequeños, con CP y Tucker como los dos extremos de una misma familia.
